```python
sector_expenditure = df.groupby("Sector")["Expenditure(million USD)"].sum()
sector_expenditure
```

```python
plt.figure(figsize=(10,5))
plt.plot(yearly_expenditure.index, yearly_expenditure.values)
plt.title("Yearly Expenditure")
plt.xlabel("Year")
plt.ylabel("Expenditure (million USD)")
plt.show()
```

```python
top10 = country_expenditure.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.bar(top10.index, top10.values)
plt.title("Top 10 Countries by Expenditure")
plt.xlabel("Country")
plt.ylabel("Expenditure (million USD)")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sector_expenditure.sort_values().plot(kind="barh")
plt.title("Expenditure by Sector")
plt.xlabel("Expenditure (million USD)")
plt.ylabel("Sector")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["GDP(%)"], df["Expenditure(million USD)"])
plt.title("Expenditure vs GDP")
plt.xlabel("GDP (%)")
plt.ylabel("Expenditure (million USD)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Expenditure(million USD)"].dropna(), bins=30)
plt.title("Distribution of Expenditure")
plt.xlabel("Expenditure (million USD)")
plt.ylabel("Frequency")
plt.show()
```

```python
avg_gdp_sector = df.groupby("Sector")["GDP(%)"].mean()

plt.figure(figsize=(10,6))
avg_gdp_sector.sort_values().plot(kind="barh")
plt.title("Average GDP Percentage by Sector")
plt.xlabel("Average GDP (%)")
plt.ylabel("Sector")
plt.show()
```
